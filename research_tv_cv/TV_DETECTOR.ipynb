```python
from ultralytics import YOLO
import cv2

# Загрузка предобученной модели YOLOv8
model = YOLO('yolov8n.pt')  # 'n' — nano версия (самая легкая)

```

```python
# Загрузка изображения
image = cv2.imread('./data/tv_data/tv1.png')


# Детекция объектов
results = model(image)

# Инициализация счетчика телевизоров
tv_count = 0

# Перебираем все обнаруженные объекты
for result in results:
    for box in result.boxes:
        class_id = int(box.cls)
        if class_id == 62:  # Класс TV/Monitor в COCO
            tv_count += 1

# Визуализация результатов с bounding boxes
annotated_image = results[0].plot()  # Автоматически рисует все обнаруженные объекты

# Добавляем текст с количеством телевизоров
text = f'TV Count: {tv_count}'
cv2.putText(annotated_image, text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

# Сохраняем результат
cv2.imwrite('./data/tv_data/detected_tvs/detected_tvs.jpg', annotated_image)

# Показываем результат в окне
cv2.imshow('TV Detection', annotated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Выводим количество в консоль
print(f"Найдено телевизоров: {tv_count}")
```

```python

# Загрузка видео
video_path = "./data/tv_data/tv_video1.mp4"
cap = cv2.VideoCapture(video_path)

# Получение параметров видео
fps = int(cap.get(cv2.CAP_PROP_FPS))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Создание VideoWriter для сохранения результата
output_path = "./data/tv_data/detected_tvs/output_video.mp4"
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

# Обработка каждого кадра
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Детекция объектов
    results = model(frame, classes=[62])  # Фильтр только для телевизоров

    # Подсчет телевизоров
    tv_count = len(results[0].boxes)

    # Визуализация
    annotated_frame = results[0].plot()
    cv2.putText(annotated_frame, f"TVs: {tv_count}", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    # Запись кадра в выходное видео
    out.write(annotated_frame)

    # Показ в реальном времени (опционально)
    cv2.imshow("TV Detection", annotated_frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Освобождение ресурсов
cap.release()
out.release()
cv2.destroyAllWindows()

print("Обработка видео завершена!")
```
